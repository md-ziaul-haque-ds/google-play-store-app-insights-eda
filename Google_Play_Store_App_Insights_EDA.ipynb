Step 15.3.2 — Convert Size to numeric MB

```python
# Replacing device-dependent sizes with missing values
df['Size'] = df['Size'].replace('Varies with device', np.nan)

# Converting megabyte values into numeric MB
df['Size'] = df['Size'].str.replace('M', '', regex=False)

# Converting kilobyte values into MB
df['Size'] = df['Size'].str.replace('k', '', regex=False)

# Converting the cleaned size values into numeric format
df['Size'] = pd.to_numeric(df['Size'], errors='coerce')

# Converting kilobytes into megabytes
df.loc[df['Size'].notna() & df_raw['Size'].reindex(df.index).str.endswith('k', na=False), 'Size'] /= 1000
```

Step 15.3 — Final, corrected Size conversion

```python
# Mapping the current DataFrame indices to their original CSV indices
original_indices = df.index + (df.index >= 10472)

# Retrieving the original Size values from the backup
original_size = df_raw.loc[original_indices, 'Size'].set_axis(df.index)

# Removing the M and k suffixes
size_mb = original_size.str.replace('M', '', regex=False)
size_mb = size_mb.str.replace('k', '', regex=False)

# Converting the cleaned values into numeric format
size_mb = pd.to_numeric(size_mb, errors='coerce')

# Identifying values originally recorded in kilobytes
kb_mask = original_size.str.endswith('k', na=False)

# Converting kilobytes into megabytes
size_mb.loc[kb_mask] = size_mb.loc[kb_mask] / 1000

# Updating the Size column
df['Size'] = size_mb
```

```python
# Checking the Size data type
print("Data type:", df['Size'].dtype)

# Checking the total missing values
print("Missing values:", df['Size'].isna().sum())

# Displaying sample converted values
print("\nSample values:")
print(df['Size'].head(15))

# Checking the minimum and maximum app sizes
print("\nMinimum size:", df['Size'].min())
print("Maximum size:", df['Size'].max())
```

Step 17.3 — Create Install Groups

```python
# Defining install group boundaries
install_bins = [
    0, 10000, 100000, 1000000, 10000000, float('inf')
]

# Defining labels for each install group
install_labels = [
    'Low', 'Moderate', 'High', 'Very High', 'Massive'
]

# Categorizing apps based on their install counts
df['Install Group'] = pd.cut(
    df['Installs'],
    bins=install_bins,
    labels=install_labels,
    right=False
)

# Checking the new feature
print(df[['Installs', 'Install Group']].head(15))

# Checking the number of apps in each group
print("\nInstall Group Distribution:")
print(df['Install Group'].value_counts().sort_index())
```

Step 17.4 — Create Price Groups

```python
# Defining price group boundaries
price_bins = [0, 0.01, 2, 5, 10, float('inf')]

# Defining labels for each price group
price_labels = ['Free', 'Budget', 'Standard', 'Premium', 'High-Priced']

# Categorizing apps based on their prices
df['Price Group'] = pd.cut(
    df['Price'],
    bins=price_bins,
    labels=price_labels,
    right=False
)

# Displaying sample price groups
print(df[['Price', 'Price Group']].head(15))

# Checking the distribution of price groups
print("\nPrice Group Distribution:")
print(df['Price Group'].value_counts().sort_index())
```

# 2. Important business insights
Reviews vs Installs (0.63)
- Apps with more installs generally tend to receive more reviews.
- This is the strongest positive relationship in the matrix.
Rating vs Installs (0.05)

- Popularity and user satisfaction are different measures.
Rating vs Price (-0.02)
- App pricing has almost no linear relationship with ratings.
Size vs Installs (0.17)
- App size has only a weak relationship with installs.
# 3. Important statistical concept
Correlation measures the strength and direction of a linear relationship.
- Close to +1: Strong positive relationship.
- Close to -1: Strong negative relationship.
- Close to 0: Little or no linear relationship.
Remember: Correlation does not establish causation.

# Step 18.10 — Advanced Analysis
# Do highly installed apps also receive higher ratings and more reviews?

Group-wise performance analysis

```python
# Comparing app performance across install groups
install_performance = df.groupby(
    'Install Group',
    observed=True
).agg(
    App_Count=('App', 'count'),
    Average_Rating=('Rating', 'mean'),
    Average_Reviews=('Reviews', 'mean')
)

# Displaying the performance summary
print(install_performance.round(2))
```

# Business insights
Finding 1 — Popularity and engagement

Average reviews increase dramatically with install groups. Massive-install apps receive approximately 2.16 million reviews on average.

Finding 2 — User satisfaction

Massive-install apps have the highest average rating (4.34), while Moderate-install apps have the lowest (4.04).

Finding 3 — Popularity doesn't guarantee satisfaction

Low-install apps average 4.16, which is higher than Moderate-install apps at 4.04.
# Overall conclusion
There is a clear relationship between app popularity and review volume. However, ratings don't increase consistently across every install group.

# Business recommendation:
Developers should focus not only on acquiring downloads but also on improving user experience and encouraging meaningful user engagement.
